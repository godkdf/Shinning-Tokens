One **1,200-second (20-minute)** training attempt on **one TPU v5e** with a
**256 MiB original UTF-8 corpus**, **16 human languages**, and six code/markup
buckets. Run cells top to bottom. Earlier notebooks and corpus caches stay intact.

Start at **1 × 1,024 for the first 100 updates**. At every **100 official
updates**, inspect timing and memory. Increase short batch by **one sequence**
when the next executable fits **85% of reported device capacity**, including
the inactive context's resident inputs and fixed evaluation buffers. The
remaining **15%** is headroom, not a guarantee against every runtime allocation.

Each increased batch must deliver at least **2% more useful tokens/s** over
its next **100 short-context training updates**. Otherwise return to the prior
batch and hold. Completed updates, weights and optimizer moments are retained.
There are **zero throwaway batch-probe training updates**. The numeric safety
cap is **32 sequences**; reaching it does not establish a global optimum.
Checks continue every 100 updates after holding. Long-context checks defer
short-batch decisions until enough short-context updates are available.

Once short batch reaches **3**, use cumulative **scored training tokens**:

| Repeating interval | Training shape |
|---|---|
| 0–3 million tokens | Current accepted short batch × 1,024 |
| 3–4 million tokens | 1 × 2,048 |
| 4–7 million tokens | Current accepted short batch × 1,024 |
| 7–8 million tokens | 1 × 2,048 |

Repeat every **4 million tokens**. Switch immediately after the first complete
optimizer update crossing a boundary. Overshoot is less than one update's
scored tokens. Do not cut gradients, discard targets or reset the data cursor.
The two packings have independent persistent row cursors over the same text;
switching back resumes that packing's next row. This intentionally revisits
the same corpus under different contexts; reported exposure includes repeats.

Initial short/long compilation and evaluation occur before the training timer.
**Every later candidate compilation counts inside the 1,200-second wall-clock
budget**, and its time is also reported separately. Do not initiate a candidate
compile with less than **60 seconds** remaining. An in-flight compile can
overrun the deadline; record it and start no further optimizer update.

Preserve the V3.8 native dense exact full-vocabulary loss, AdamW, learning rate
**0.0003**, **5 warmup updates then constant**, BF16 compute, FP32 master
weights, gradient clipping **1.0**, decay **0.01**, and tied embeddings. Default
model: **268,100,736 random parameters**. Batch changes do not rescale the LR.

# V3.9 engineering contract

- One chip, one initialization, one attempt, one 1,200-second interval.
- First 100 updates: 1 × 1,024; checks land exactly on multiples of 100 updates.
- At most 16 official updates per dispatch. Never discard a completed update.
- Increase short batch by one after memory admission, using real training
  windows for timing. A <2% improvement rolls the shape back one and ends the
  ramp. No weights, Adam moments, step count or learning-rate reset.
- Memory gate: compiler arguments + outputs − aliases + temporaries, plus
  inactive context and evaluation inputs, ≤85% of device capacity. Gate the
  next compilation with a conservative extrapolation first. Missing estimates
  or capacity stop increases. Do not intentionally provoke OOM or retry training.
- Context schedule follows cumulative actual scored tokens, including EOS,
  excluding padding. Enable long context only while accepted short batch ≥3.
  Complete the crossing update and record the exact transition token count.
- Separate row cursors preserve next-row continuity across shapes and modes.
  Gather cyclically without batch-size padding or repeated first rows on switches.
- Expanded source quotas sum to 10,000 basis points of original UTF-8 bytes.
  Freeze revisions and raw hashes on first preparation; share the new corpus
  across later tokenizers. Repository-based code splitting prevents files from
  the same identified repository crossing train/evaluation boundaries.
- Prepare both context caches with distinct keys and checksums. Fixed scoring
  uses only the 1,024-context holdout, 64 KiB per language plus 1 MiB in-mixture.
- CPU prefix accounting must match device token totals after every dispatch,
  and final byte totals across both contexts. Accounting arrays stay on CPU.
- Finite checks reject a bad update; no automatic attempt or batch retry.
- Preserve model, exact dense loss, optimizer, precision and initialization.
- Save official evaluation and FP32 weights before diagnostic updates. Verify
  file hashes/schema without loading weights back into a model.
- Record allocator memory samples, reported capacity, compiler estimates by
  shape, total memory estimates including both resident contexts, batch checks,
  context switches, compilation timing, useful tokens/s, bytes/s and CU estimates.
- Request matrix-unit and HBM counters, host/device idle and operation timings.
  Unsupported/absent measurements are unavailable and fail profiling
  completeness; cost-model estimates must never fill measured-counter fields.
- Profiling: 16 final-short updates +16 long updates; mixed operation averages
  explicitly labelled. Diagnostic tokens and weights excluded from official
  results, but diagnostic time included in total runtime/cost.
- Separate engineering, learning, performance and profiling statuses. Throughput
  ≥45,734 tokens/s, pretraining compilation ≤180 s, total report ≤1,800 s and
  useful occupancy ≥95% are targets, not promises. Preparation may exceed them.
- Builders exclusively create V3.9 files and refuse to overwrite delivered files.

French was already present. Russian, Japanese, German and Igbo expand the
human-language list from **12 to 16**, including English. Translation buckets
remain French/Arabic/Hindi/Swahili at **1.25% each**, containing both the original
and English side, with educational score ≥2 and no early-stop rows. Actual
language shares include English translations and are reported. Code uses six
distinct domain labels; its `eng_Latn` language ID is an accounting placeholder,
not a claim that comments or strings are all English. These are declared
engineering allocations, not measured optimal mixture weights.

Holdout: **64 KiB ×16 FineWiki languages =1 MiB**, plus **1 MiB** in-mixture
allocated by the source percentages. FineWiki configurations are checked at the
pinned dataset revision; unavailable configurations or byte shortages stop
preparation. No fabricated examples, language replacement or quota renormalizing.

Freeze source commits once per new shared recipe. Preserve exact raw byte
hashes, provenance, deterministic group splits and global exact source/span
deduplication. Code repository names from both source schemas are grouped
together. Near duplicates and unidentified cross-repository copies are not
exhaustively removed. Spans are exact UTF-8 prefixes up to **32 KiB**. Bounded
stream-prefix selection is not a uniform sample of the full source collection.

Use at most **6 concurrent source jobs**; persist source batches in Drive.
Initial candidate target is 125% of quota, at least four spans per stratum;
continue a pinned stream if deduplication consumes that buffer. The per-job
candidate JSONL cap remains 256 MiB. Stack-smol contains a finite 10,000 samples
per language: quota fulfilment is checked, not assumed. Source access and
provenance terms are those of the upstream datasets.

Both 1,024 and 2,048 packings preserve all original text targets, independent
virtual EOS accounting, and document isolation. Each packing has its own cache
key and row order. Both reside on TPU; CPU-only byte/source prefixes are used
to account for the exact rows consumed, including repeat passes. A context
switch resumes its saved row cursor. It does not advance a single common raw
text frontier, so examples may recur across the two packings before either has
completed a full pass. Report context-specific exposure, not unique data seen.

BPB excludes virtual EOS and divides original-text negative log likelihood by
original UTF-8 bytes ×ln(2). Fixed 1,024-token evaluation scores the same held-out
text before/after training. Report 16-language macro BPB plus the original
12-language subset. Expanded recipe, potentially different source revisions,
and longer training prevent a controlled comparison with V3.5/V3.8.
